```python
def cleaing(text):
    text_bersih = text.lower()
    text_bersih = re.sub(r'https\S+', '', text_bersih)
    text_bersih = re.sub(r'www\.\S+', '', text_bersih)
    text_bersih = re.sub(r'@[A-Za-z0-9_]+', '', text_bersih)
    text_bersih = re.sub(r'#([A-Za-z0-9_])+', '', text_bersih)
    text_bersih = re.sub(r'[^a-z\s]', '', text_bersih)
    text_bersih = re.sub(r'\d+', '', text_bersih)
    text_bersih = re.sub(r'\s+', ' ', text_bersih)

    return text_bersih

tes = "tweet_alay = Haloo!! @admin saya KECEWA 0.9 999 banget 😡. Barang rusak! Refund donk rp 50.000!! Cek https://tokoadmin.com #parah"

hasil=cleaing(tes)
print(hasil)
```
